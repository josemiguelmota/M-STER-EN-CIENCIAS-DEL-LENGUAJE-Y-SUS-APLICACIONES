## 2. ¿Subcadena o palabra completa?

Predice el número que sale. Señala en qué partes de la frase aparece `el`. Cambia una línea para contar solo la palabra completa `el` en esta frase, que no contiene puntuación. Comprueba después con `El elefante`.

**Explicación:** …

```python
frase = "El elefante mira el papel"
# lower() devuelve una copia del texto en minúsculas.
normalizada = frase.lower()
# count() aplicado a un texto cuenta apariciones de una subcadena.
recuento = normalizada.count("el")
# print() muestra el recuento en la salida de la celda.
print(recuento)
```

## 4. ¿Cómo se construye el recuento?

Sigue a mano las primeras cuatro vueltas del bucle: ¿qué claves hay en el diccionario y cuánto valen? Predice la salida final. Después cambia el primer `el` a `El`: ¿aparece otra clave? Modifica el bucle para que las dos formas cuenten juntas y actualiza el comentario.

**Predicción:** …  
**Explicación:** …  
**Cambio y resultado:** …

```python
texto = "el gato y el perro y el ratón"
frecuencias = {}
# split() separa el texto para que el bucle recorra una forma cada vez.
for forma in texto.split():
    if forma not in frecuencias:
        frecuencias[forma] = 1
    else:
        frecuencias[forma] += 1
# print() muestra el diccionario con todas las frecuencias.
print(frecuencias)
```

```python
# contar_palabras() recibe un texto y devuelve el número de formas separadas por espacios en blanco.
def contar_palabras(texto):
    # split() devuelve la lista de formas y len() cuenta sus elementos.
    return len(texto.split())

# La llamada devuelve el recuento y lo guardamos en una variable.
total = contar_palabras("La lengua cambia")
# print() muestra el valor guardado; no forma parte del cálculo.
print(total)
```
